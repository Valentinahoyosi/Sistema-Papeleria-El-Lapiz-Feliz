## Paso 6 — opción 1: registrar una factura con `for`

Completa la opción `1` del menú. Necesitas, en este orden:

1. Preguntar cuántos productos tiene la factura (`cantidad`, convertida a entero con `int()`)
2. Crear `total_factura` empezando en 0
3. Un `for i in range(cantidad):` que en cada vuelta pida nombre y precio, y sume el precio a `total_factura`
4. Imprimir el total de la factura
5. Sumar `total_factura` a `total_dia`
6. Sumar 1 a `facturas_hechas`

Recuerda poner `total_dia = 0` y `facturas_hechas = 0` **antes** del `while` (paso 5), no adentro de la opción 1.

````python
total_dia = 0
facturas_hechas = 0
tarifa_IVA = 0.19

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Reiniciar el total del día")
    print("5. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad_productos = int(input("¿Cuántos productos diferentes tiene esta factura? "))
        subtotal_factura = 0

        for i in range(cantidad_productos):
            print(f"Producto {i + 1} de {cantidad_productos}:")
            nombre = input("  Nombre del producto: ")
            cantidad_unidades = int(input("Cantidad: "))
            valor_unitario = float(input("Valor unitario: "))

            subtotal_producto = cantidad_unidades * valor_unitario
            subtotal_factura = subtotal_factura + subtotal_producto

        descuento = 0
        if subtotal_factura > 50000:
            descuento = subtotal_factura * 0.05
            print(f"¡Descuento aplicado! -${descuento:,.0f}")

        subtotal_con_descuento = subtotal_factura - descuento
        iva = subtotal_con_descuento * tarifa_IVA
        total_factura = subtotal_con_descuento + iva

        print("-" * 40)
        print(f"Subtotal:  ${subtotal_factura:,.0f}")
        if descuento > 0:
            print(f"Descuento: -${descuento:,.0f}")
        print(f"IVA (19%):      ${iva:,.0f}")
        print(f"Total a pagar : ${total_factura:,.0f}")
        print("-" * 40)

        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1

    elif opcion == "2":
        print(f"El total acumulado del día es: ${total_dia:,.0f}")
    elif opcion == "3":
        print(f"Se han registrado {facturas_hechas} facturas hoy.")
    elif opcion == "4":
        total_dia = 0
        facturas_hechas = 0
        print("¡Día reiniciado! El total y el contador de facturas volvieron a 0.")
    elif opcion == "5":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")

````

## Paso 7 — opciones 2 y 3: mostrar lo acumulado

Completa las opciones `2` y `3`, mostrando `total_dia` y `facturas_hechas`. Esta es la versión final y completa del menú.

````python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))
        total_factura = 0
        for i in range(cantidad):
            print(f"Producto {i+1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio
        print(f"Factura registrada. Total : ${total_factura:,.0f}")
        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1
    elif opcion == "2":
        print(f"El total acumulado del día es: ${total_dia:,.0f}")
    elif opcion == "3":
        print(f"Se han registrado {facturas_hechas} facturas en total.")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````

## Paso 8 - Proyecto integrador: Sistema de Facturación de la Papelería

Con todo lo aprendido (**variables, operadores, condicionales, ciclos, listas/diccionarios y funciones**), se arma el sistema completo.

**Se construye en Python**, que es el lenguaje más práctico para un programa interactivo de este tipo (input del usuario, generación de factura en texto). Al final verás cómo se vería la versión de cálculo equivalente en R.



````python
# ==========================================
# PROYECTO: EL LÁPIZ FELIZ - SISTEMA INTERACTIVO
# ==========================================

print("-" * 48)
print("PAPELERÍA EL LÁPIZ FELIZ - SISTEMA INTERACTIVO")
print("-" * 48)

# Catálogo de productos
catalogo = {
    "P-001": {"nombre": "Cuaderno cuadriculado", "precio": 8500, "iva": 0.19},
    "E-011": {"nombre": "Caja de lapiceros x12", "precio": 15990, "iva": 0.19},
    "L-030": {"nombre": "Libro juvenil", "precio": 45000, "iva": 0.0},
    "R-050": {"nombre": "Revista educativa", "precio": 12000, "iva": 0.05}
}

# Catálogo de clientes
clientes = {
    "C-01": {"nombre": "Doña Luz", "tipo": "detal"},
    "C-02": {"nombre": "Colegio San Andrés", "tipo": "colegio"},
    "C-03": {"nombre": "Oficina Contadores S.A.S.", "tipo": "oficina"}
}

# Inventario actual en bodega
inventario = {"P-001": 40, "E-011": 12, "L-030": 5, "R-050": 20}

# Mostrar catálogo disponible para el cajero
print("\nCATÁLOGO DISPONIBLE:")
for codigo, p in catalogo.items():
    print(f"[{codigo}] {p['nombre']} - ${p['precio']} (IVA: {int(p['iva']*100)}%)")
print("-" * 48)

print("--- CONSULTA DE INVENTARIO ---")
codigo = input("Código a consultar: ").strip().upper()
if codigo in inventario:
    print(f"{catalogo[codigo]['nombre']} → {inventario[codigo]} unidades")
else:
    print("Ese código no existe en la papelería")
print("-" * 48)

# Mostrar la información de los clientes en pantalla
print("DIRECTORIO DE CLIENTES:")
for doc, info in clientes.items():
    print(f"[{doc}] Nombre: {info['nombre']} | Tipo: {info['tipo']}")
print("-" * 48)

# Datos fijos de la factura (Tupla)
encabezado = ("F-102", "08/08/2026", "C-02")
print(f"Factura N°: {encabezado[0]} | Fecha: {encabezado[1]}")

# Seleccionar cliente
nit_cliente = input("Ingrese el código del cliente (C-01, C-02, C-03): ").strip().upper()
if nit_cliente not in clientes:
    nit_cliente = "C-01"  # Por defecto si no existe
    print("Cliente no encontrado. Se asignó por defecto: C-01 (Doña Luz)")

nombre_cliente = clientes[nit_cliente]["nombre"]
tipo_cliente = clientes[nit_cliente]["tipo"]
print(f"Cliente seleccionado: {nombre_cliente} ({tipo_cliente.upper()})")
print("-" * 48)

# Ciclo interactivo para tomar el pedido (Misión 3 y 4)
pedido = []
print("Escribe el código del producto para agregarlo. Escribe 'fin' para terminar.")

while True:
    codigo_ingresado = input("Código de producto (o 'fin'): ").strip().upper()

    if codigo_ingresado == "FIN":
        break

    if codigo_ingresado not in catalogo:
        print("Ese código no existe en la papelería. Intenta de nuevo.")
        continue

    cantidad_str = input(f"  Cantidad para '{catalogo[codigo_ingresado]['nombre']}': ").strip()

    if not cantidad_str.isdigit() or int(cantidad_str) <= 0:
        print("Cantidad inválida. Debe ser un número mayor a 0.")
        continue

    cantidad = int(cantidad_str)

    # Validar stock disponible en inventario
    stock_actual = inventario.get(codigo_ingresado, 0)
    if cantidad > stock_actual:
        print(f"Stock insuficiente. Solo hay {stock_actual} unidades disponibles.")
        continue

    # Guardar en el pedido
    pedido.append((codigo_ingresado, cantidad))
    print(f"Agregado con éxito: {cantidad} unidades.")

# Procesar factura si hay productos
if len(pedido) > 0:
    total = 0
    print("\n" + "=" * 45)
    print("DETALLE DE FACTURA")
    print("=" * 48)

    for codigo, cantidad in pedido:
        p = catalogo[codigo]
        subtotal = p["precio"] * cantidad
        iva = subtotal * p["iva"]
        total_linea = subtotal + iva
        total += total_linea
        print(f"{p['nombre']} (x{cantidad}) -> Subtotal con IVA: ${total_linea:.2f}")

    # Aplicar descuento institucional si es colegio y supera los 200.000
    if tipo_cliente == "colegio" and total >= 200000:
        total = total * 0.9
        print("\n[¡Descuento del 10% aplicado por convenio escolar!]")

    print("-" * 48)
    print(f"TOTAL A PAGAR: ${total:.2f}")
    print("=" * 48)

    # Descontar automáticamente lo facturado del inventario
    for codigo, cantidad in pedido:
        inventario[codigo] -= cantidad

    print("\nInventario actualizado en bodega:")
    for codigo, stock in inventario.items():
        print(f"[{codigo}] {catalogo[codigo]['nombre']} -> Quedan: {stock} unidades")
else:
    print("\nNo se registraron productos en la factura.")
````
